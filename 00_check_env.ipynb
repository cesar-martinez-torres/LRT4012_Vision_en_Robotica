# Environment Check

Run the following cell after selecting the **Python (Robot Vision 2026)** kernel. The test prints the installed package versions and generates a simple image using NumPy, OpenCV, and Matplotlib.

```python
import sys

import cv2
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy
import skimage
from PIL import Image

versions = {
    "Python": sys.version.split()[0],
    "OpenCV": cv2.__version__,
    "NumPy": np.__version__,
    "SciPy": scipy.__version__,
    "pandas": pd.__version__,
    "Matplotlib": matplotlib.__version__,
    "scikit-image": skimage.__version__,
    "Pillow": Image.__version__,
}

for package, version in versions.items():
    print(f"{package:14s}: {version}")

image_bgr = np.zeros((240, 400, 3), dtype=np.uint8)
cv2.rectangle(image_bgr, (40, 40), (360, 200), (0, 180, 0), 4)
cv2.putText(
    image_bgr,
    "Environment OK",
    (72, 132),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.9,
    (255, 255, 255),
    2,
    cv2.LINE_AA,
)

image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(8, 4.8))
plt.imshow(image_rgb)
plt.title("Robot Vision environment test")
plt.axis("off")
plt.show()

print("\nAll environment checks completed successfully.")
```
