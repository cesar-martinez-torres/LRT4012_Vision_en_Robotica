
# 📘 Introduction to OpenCV — Loading and Displaying Images

This notebook shows how to:

1. Read an image from a file with **OpenCV** (`cv2.imread`).
2. Display it using **Matplotlib**.
3. Convert from BGR (OpenCV's internal format) to RGB to display colors correctly.
4. Save a processed image in the student's image directory

```python

# Import the required libraries.
from pathlib import Path

import cv2
import matplotlib.pyplot as plt

# Identify the repository root from the current working directory.
search_roots = [Path.cwd(), *Path.cwd().parents]
repository_root = next(
    (
        root
        for root in search_roots
        if (root / "requirements.txt").is_file()
        and (root / "imagenes").is_dir()
        and (root / "student_work").is_dir()
    ),
    None,
)

if repository_root is None:
    raise FileNotFoundError("The root of the Vision_en_Robotica repository could not be located.")

COURSE_IMAGES = repository_root / "imagenes"
STUDENT_IMAGES = repository_root / "student_work" / "imagenes"
image_path = COURSE_IMAGES / "lenna.png"

# Read the official course image in BGR channel order.
img_bgr = cv2.imread(str(image_path), cv2.IMREAD_COLOR)
if img_bgr is None:
    raise FileNotFoundError(f"OpenCV could not load the course image: {image_path}")

print("Image loaded successfully with size:", img_bgr.shape)

plt.imshow(img_bgr)
plt.title("Displayed as BGR")
plt.axis("off")
plt.show()

print(img_bgr) #complete array
px_bgr=img_bgr[100,100]
print("BGR:",px_bgr)  # Print the pixel value at position (100, 100)

```

```python

# ✅ Display the image with Matplotlib
# OpenCV loads images in BGR format; convert to RGB to display them correctly
if img_bgr is not None:
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    plt.imshow(img_rgb)
    plt.title("Image loaded with OpenCV")
    plt.axis("off")   # hide axes
    plt.show()

# ✅ Print the image NumPy array
px_rgb=img_rgb[100,100]
print("RGB:",px_rgb)  # Print the pixel value at position (100, 100)
```

```python

# ✅ Convert to grayscale
if img_bgr is not None:
    img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
    plt.imshow(img_gray, cmap="gray")
    plt.title("Grayscale image")
    plt.axis("off")
    plt.show()

#print(img_gray)
px_gray=img_gray[100,100]
print("Gray:",px_gray)  # Print the pixel value at position (100, 100)
```

```python

# Save student-generated files outside the official course image directory.
STUDENT_IMAGES.mkdir(parents=True, exist_ok=True)
output_path = STUDENT_IMAGES / "lenna_gray.png"

if not cv2.imwrite(str(output_path), img_gray):
    raise IOError(f"OpenCV could not save the image: {output_path}")

print(f"Grayscale image saved in: {output_path}")
```
